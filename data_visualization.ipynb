```python
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May"]
sales = [200,250,300,280,350]

plt.plot(months, sales, marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
days = [1,2,3,4,5]
temp = [30,32,31,29,33]

plt.plot(days,temp,color='red')
plt.title("Temperature Change")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.show()
```

```python
products = ["A","B","C","D"]
sales = [50,70,30,90]

plt.bar(products,sales)
plt.title("Product Sales Comparison")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
students = ["John","Sam","Ravi","Anu"]
marks = [80,75,90,85]

plt.bar(students,marks,color='green')
plt.title("Student Marks")
plt.show()
```

```python
x = [1,2,3,4]
y1 = [10,20,25,30]
y2 = [15,18,28,35]

plt.plot(x,y1,label="Product A")
plt.plot(x,y2,label="Product B")

plt.legend()
plt.title("Product Growth")
plt.show()
```

```python
scores = [50,60,70,80,90,85,75,65,55,95]

plt.hist(scores,bins=5)
plt.title("Exam Score Distribution")
plt.show()
```

```python
ages = [18,19,20,21,22,23,24,25,21,22]

plt.hist(ages,bins=4,color='orange')
plt.title("Age Distribution")
plt.show()
```

```python
height = [150,160,165,170,175]
weight = [50,60,65,70,75]

plt.scatter(height,weight)
plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Height vs Weight")
plt.show()
```

```python
x = [10,20,30,40]
y = [15,25,35,45]
sizes = [50,100,150,200]

plt.scatter(x,y,s=sizes,c='green')
plt.title("Scatter with Size Variation")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = [10,20,20,30,40,40,50]
sns.histplot(data)
plt.show()
```

```python
sns.kdeplot(data)
plt.show()
```

```python
tips = sns.load_dataset("tips")
sns.histplot(tips["total_bill"])
plt.show()
```

```python
sns.histplot(data=[tips["total_bill"],tips["tip"]])
plt.show()
```

```python
sns.set_style("darkgrid")
sns.histplot(tips["total_bill"],color="purple")
plt.show()
```

```python
sns.boxplot(x=[10,20,30,40,100])
plt.show()
```

```python
tips = sns.load_dataset("tips")
sns.boxplot(x="day",y="total_bill",data=tips)
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
"Age":[25,30,35,40,45],
"Income":[20000,30000,40000,50000,60000],
"Spending":[200,300,400,450,500]
}

df = pd.DataFrame(data)

sns.scatterplot(x="Age",y="Income",data=df)
plt.show()

sns.heatmap(df.corr(),annot=True)
plt.show()

sns.boxplot(data=df)
plt.show()
```
